```python
regions_expenses = df.groupby("region")["expenses"].agg(mean="mean")

regions_expenses
```

```python
#data visualizations

import matplotlib.pyplot as plt

plt.bar(regions_expenses.index,regions_expenses["mean"],edgecolor="red",alpha=0.7)
plt.title("Expenses per regions")
plt.xlabel("Regions")
plt.ylabel("Expenses")
plt.grid()
plt.show()
plt.savepng
```
